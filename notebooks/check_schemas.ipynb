# Database Schema Check

Inspect schemas, tables, and models in the DuckDB database.

```python
import duckdb
from pathlib import Path

db_path = Path.cwd().parent / 'lebron_analytics.duckdb'
conn = duckdb.connect(str(db_path))
print(f'Connected: {db_path}')
```

## All Schemas

```python
result = conn.execute("""
	SELECT schema_name
	FROM information_schema.schemata
	ORDER BY schema_name
""").fetchall()

for schema in result:
	print(f'  - {schema[0]}')
```

## All Tables by Schema

```python
result = conn.execute("""
	SELECT table_schema, table_name, table_type
	FROM information_schema.tables
	WHERE table_schema NOT IN ('information_schema', 'pg_catalog')
	ORDER BY table_schema, table_name
""").fetchall()

current_schema = None
for schema, table, ttype in result:
	if schema != current_schema:
		print(f'\n{schema}:')
		current_schema = schema
	print(f'  - {table} ({ttype})')
```

## Intermediate Models (int_*)

```python
result = conn.execute("""
	SELECT table_schema, table_name
	FROM information_schema.tables
	WHERE table_name LIKE 'int_%'
	ORDER BY table_schema, table_name
""").fetchall()

if result:
	for schema, table in result:
		print(f'  {schema}.{table}')
else:
	print('  none found')
```

## Staging Models (stg_*)

```python
result = conn.execute("""
	SELECT table_schema, table_name
	FROM information_schema.tables
	WHERE table_name LIKE 'stg_%'
	ORDER BY table_schema, table_name
""").fetchall()

if result:
	for schema, table in result:
		print(f'  {schema}.{table}')
else:
	print('  none found')

conn.close()
```
